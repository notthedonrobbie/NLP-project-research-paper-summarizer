```python
from google.colab import files
from research_assistant import ResearchAssistant

uploaded = files.upload()
filename = next(name for name in uploaded if name.lower().endswith('.pdf'))
assistant = ResearchAssistant.from_pdf(uploaded[filename], title=filename, retrieval_method='bm25')
print(f'Pages: {assistant.document.page_count}; chunks: {len(assistant.chunks)}')
```

```python
summary = assistant.summarize(method='textrank')
print('OVERVIEW\n', summary.overview)
for field, text in summary.sections.items():
    print(f'\n{field.upper()}\n{text}')
```

```python
question = 'What dataset was used?'
answer = assistant.ask(question, top_k=5)
print(answer.text)
for citation in answer.citations:
    print(f'\nPages {citation.chunk.page_numbers}:\n{citation.chunk.text}')
```
